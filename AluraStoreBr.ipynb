```python
import pandas as pd

# Carregar os dados
url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"


loja1 = pd.read_csv(url)
loja2 = pd.read_csv(url2)
loja3 = pd.read_csv(url3)
loja4 = pd.read_csv(url4)

# Calcular o faturamento total de cada loja
faturamento_loja1 = loja1['Preço'].sum()
faturamento_loja2 = loja2['Preço'].sum()
faturamento_loja3 = loja3['Preço'].sum()
faturamento_loja4 = loja4['Preço'].sum()

# Criar um DataFrame para exibir os resultados
faturamento_total = pd.DataFrame({
    'Loja': ['Loja 1', 'Loja 2', 'Loja 3', 'Loja 4'],
    'Faturamento Total': [faturamento_loja1, faturamento_loja2, faturamento_loja3, faturamento_loja4]
})

print(faturamento_total)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Carregar os dados
urls = {
    "Loja 1": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv",
    "Loja 2": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv",
    "Loja 3": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv",
    "Loja 4": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"
}

# Dicionário para armazenar os faturamentos
faturamento_por_loja = {}

for nome_loja, url in urls.items():
    df = pd.read_csv(url)
    if 'Preço' in df.columns:  # Verifica se a coluna de preço existe
        faturamento = df['Preço'].sum()
        faturamento_por_loja[nome_loja] = faturamento

# Criar DataFrame com os resultados
df_faturamento = pd.DataFrame(list(faturamento_por_loja.items()), columns=['Loja', 'Faturamento'])

# Configurar o gráfico
plt.figure(figsize=(10, 6))  # Tamanho do gráfico
bars = plt.bar(df_faturamento['Loja'], df_faturamento['Faturamento'], color='skyblue')

# Adicionar rótulos e título
plt.title('Faturamento por Loja', fontsize=16)
plt.xlabel('Lojas', fontsize=12)
plt.ylabel('Faturamento Total (R$)', fontsize=12)

# Adicionar valores nas barras
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height,
             f'R$ {height:,.2f}',
             ha='center', va='bottom', fontsize=10)

# Ajustar formato do eixo Y para exibir valores monetários
plt.gca().yaxis.set_major_formatter('R$ {x:,.2f}')

# Rotacionar rótulos do eixo X se necessário
plt.xticks(rotation=45, ha='right')

# Mostrar o gráfico
plt.tight_layout()  # Melhora o espaçamento
plt.show()
```

# 2. Vendas por Categoria


```python
import pandas as pd

# Carregar e limpar dados
urls = [
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv",
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv",
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv",
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"
]

lojas = [pd.read_csv(url).drop_duplicates() for url in urls]

# Função para encontrar coluna de categoria
def encontrar_coluna_categoria(df):
    possiveis_nomes = ['Categoria do Produto']
    for nome in possiveis_nomes:
        if nome in df.columns:
            return nome
    return None

# Função para contar produtos por categoria
def contar_produtos_por_categoria(df, nome_loja):
    coluna_categoria = encontrar_coluna_categoria(df)
    if coluna_categoria is None:
        print(f"Aviso: Nenhuma coluna de categoria encontrada em {nome_loja}")
        return pd.DataFrame()

    contagem = df.groupby(coluna_categoria).size().reset_index(name='Quantidade')
    contagem['Loja'] = nome_loja
    contagem = contagem.rename(columns={coluna_categoria: 'Categoria'})
    return contagem

# Processar cada loja
resultado_final = pd.concat([
    contar_produtos_por_categoria(lojas[0], 'Loja 1'),
    contar_produtos_por_categoria(lojas[1], 'Loja 2'),
    contar_produtos_por_categoria(lojas[2], 'Loja 3'),
    contar_produtos_por_categoria(lojas[3], 'Loja 4')
]).drop_duplicates()

# Obter categorias mais populares (sem repetição)
categorias_populares = (
    resultado_final
    .sort_values('Quantidade', ascending=False)
    .drop_duplicates('Loja')
    .sort_values('Loja')
)

print("Quantidade de produtos vendidos por categoria em cada loja:")
print(resultado_final.to_string(index=False))

print("\nCategorias mais populares de cada loja:")
print(categorias_populares.to_string(index=False))

```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Carregar os dados
urls = {
    "Loja 1": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv",
    "Loja 2": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv",
    "Loja 3": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv",
    "Loja 4": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"
}

# Função para contar produtos por categoria
def contar_vendas_por_categoria(df, nome_loja):
    if 'Categoria do Produto' not in df.columns:
        return pd.DataFrame()
    vendas = df['Categoria do Produto'].value_counts().reset_index()
    vendas.columns = ['Categoria do Produto', 'Vendas']
    vendas['Loja'] = nome_loja
    return vendas

# Processar todas as lojas
dados_vendas = pd.concat([contar_vendas_por_categoria(pd.read_csv(url), nome)
                    for nome, url in urls.items()])

# Pivotar para formato wide (categorias vs lojas)
pivot_vendas = dados_vendas.pivot(index='Categoria do Produto', columns='Loja', values='Vendas').fillna(0)

# Configurar o gráfico
categorias = pivot_vendas.index
lojas = pivot_vendas.columns
x = np.arange(len(categorias))
width = 0.2  # Largura das barras

fig, ax = plt.subplots(figsize=(12, 7))

# Criar barras para cada loja
for i, loja in enumerate(lojas):
    ax.bar(x + i*width, pivot_vendas[loja], width, label=loja,
           edgecolor='black', linewidth=0.7)

# Personalização
ax.set_title('Vendas por Categoria em Cada Loja', fontsize=16, pad=20)
ax.set_xlabel('Categoria', fontsize=12)
ax.set_ylabel('Número de Vendas', fontsize=12)
ax.set_xticks(x + width*(len(lojas)-1)/2)
ax.set_xticklabels(categorias, rotation=45, ha='right')
ax.legend(title='Lojas', bbox_to_anchor=(1.05, 1), loc='upper left')

# Adicionar valores nas barras
for bar in ax.containers:
    ax.bar_label(bar, fmt='%.0f', padding=2, fontsize=9)

plt.tight_layout()
plt.show()
```

# 3. Média de Avaliação das Lojas

```python
import pandas as pd

# Carregar os dados
urls = {
    "Loja 1": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv",
    "Loja 2": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv",
    "Loja 3": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv",
    "Loja 4": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"
}

# Dicionário para armazenar as médias
medias_avaliacoes = {}

# Função para detectar o nome da coluna de avaliação
def encontrar_coluna_avaliacao(df):
    possiveis_nomes = ['Avaliação da compra']
    for nome in possiveis_nomes:
        if nome in df.columns:
            return nome
    return None

# Processar cada loja
for nome_loja, url in urls.items():
    df = pd.read_csv(url).drop_duplicates()  # Remove duplicatas
    coluna_avaliacao = encontrar_coluna_avaliacao(df)

    if coluna_avaliacao is None:
        print(f"Aviso: Coluna de avaliação não encontrada em {nome_loja}")
        continue

    media = df[coluna_avaliacao].mean()
    medias_avaliacoes[nome_loja] = media

# Criar um DataFrame com os resultados
resultado = pd.DataFrame(list(medias_avaliacoes.items()), columns=['Loja', 'Média das Avaliações'])

print("\nMédia das avaliações por loja:")
print(resultado.to_string(index=False))
```

# 4. Produtos Mais e Menos Vendidos

```python
import pandas as pd

# Carregar os dados
urls = {
    "Loja 1": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv",
    "Loja 2": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv",
    "Loja 3": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv",
    "Loja 4": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"
}

# Função para encontrar a coluna de produtos
def encontrar_coluna_produto(df):
    possiveis_nomes = ['Produto', 'product', 'Item', 'Nome', 'Descrição']
    for nome in possiveis_nomes:
        if nome in df.columns:
            return nome
    return None

# Processar cada loja
resultados = []

for nome_loja, url in urls.items():
    df = pd.read_csv(url).drop_duplicates()
    coluna_produto = encontrar_coluna_produto(df)

    if coluna_produto is None:
        print(f"Aviso: Coluna de produto não encontrada em {nome_loja}")
        continue

    # Contar vendas por produto
    vendas_por_produto = df[coluna_produto].value_counts().reset_index()
    vendas_por_produto.columns = ['Produto', 'Quantidade Vendida']
    vendas_por_produto['Loja'] = nome_loja

    # Identificar mais e menos vendidos
    mais_vendido = vendas_por_produto.nlargest(1, 'Quantidade Vendida')
    menos_vendido = vendas_por_produto.nsmallest(1, 'Quantidade Vendida')

    resultados.append(mais_vendido)
    resultados.append(menos_vendido)

# Consolidar resultados
resultado_final = pd.concat(resultados).reset_index(drop=True)

# Formatar saída
print("Produtos Mais e Menos Vendidos por Loja:")
for loja in urls.keys():
    dados_loja = resultado_final[resultado_final['Loja'] == loja]
    if not dados_loja.empty:
        print(f"\n--- {loja} ---")
        print(f"→ Mais Vendido: {dados_loja.iloc[0]['Produto']} ({dados_loja.iloc[0]['Quantidade Vendida']} vendas)")
        print(f"→ Menos Vendido: {dados_loja.iloc[1]['Produto']} ({dados_loja.iloc[1]['Quantidade Vendida']} vendas)")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Carregar os dados
urls = {
    "Loja 1": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv",
    "Loja 2": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv",
    "Loja 3": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv",
    "Loja 4": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"
}

# Configurações do gráfico
plt.figure(figsize=(14, 8))
colors = ['#2ecc71', '#e74c3c']  # Verde para mais vendidos, vermelho para menos
bar_width = 0.35

# Processar cada loja
for i, (nome_loja, url) in enumerate(urls.items()):
    df = pd.read_csv(url)

    # Verificar coluna de produtos
    col_produto = 'Produto' if 'Produto' in df.columns else 'Nome' if 'Nome' in df.columns else df.columns[0]

    # Contar vendas por produto
    vendas = df[col_produto].value_counts()

    # Identificar mais e menos vendidos
    mais_vendido = vendas.idxmax(), vendas.max()
    menos_vendido = vendas.idxmin(), vendas.min()

    # Posições no gráfico
    x_pos = i * 1.5  # Espaçamento entre grupos de lojas

    # Plotar barras
    plt.bar(x_pos - bar_width/2, mais_vendido[1], width=bar_width, color=colors[0], label='Mais Vendido' if i == 0 else "")
    plt.bar(x_pos + bar_width/2, menos_vendido[1], width=bar_width, color=colors[1], label='Menos Vendido' if i == 0 else "")

    # Adicionar anotações
    plt.text(x_pos - bar_width/2, mais_vendido[1] + 5, f"{mais_vendido[0]}\n{mais_vendido[1]}", ha='center')
    plt.text(x_pos + bar_width/2, menos_vendido[1] + 5, f"{menos_vendido[0]}\n{menos_vendido[1]}", ha='center')

    # Marcação da loja
    plt.text(x_pos, -max(vendas)*0.1, nome_loja, ha='center', fontsize=12, fontweight='bold')

# Personalização do gráfico
plt.title('Produtos Mais e Menos Vendidos por Loja', fontsize=16, pad=20)
plt.ylabel('Quantidade Vendida', fontsize=12)
plt.xticks([])  # Remover ticks do eixo x pois usamos anotações
plt.legend(loc='upper right')
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Ajustar limites
current_ylim = plt.ylim()
plt.ylim(0, current_ylim[1] * 1.2)  # Espaço para as anotações

plt.tight_layout()
plt.show()
```

# 5. Frete Médio por Loja


```python
import pandas as pd

# Carregar os dados
urls = {
    "Loja 1": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv",
    "Loja 2": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv",
    "Loja 3": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv",
    "Loja 4": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"
}

# Dicionário para armazenar os fretes médios
frete_medio_por_loja = {}

# Função para detectar o nome da coluna de frete
def encontrar_coluna_frete(df):
    possiveis_nomes = ['Frete', 'frete', 'FRETE', 'Shipping', 'Custo_Frete', 'Transporte']
    for nome in possiveis_nomes:
        if nome in df.columns:
            return nome
    return None

# Processar cada loja
for nome_loja, url in urls.items():
    df = pd.read_csv(url).drop_duplicates()  # Remove duplicatas
    coluna_frete = encontrar_coluna_frete(df)

    if coluna_frete is None:
        print(f"Aviso: Coluna de frete não encontrada em {nome_loja}")
        continue

    frete_medio = df[coluna_frete].mean()
    frete_medio_por_loja[nome_loja] = frete_medio

# Criar um DataFrame com os resultados
resultado = pd.DataFrame(list(frete_medio_por_loja.items()), columns=['Loja', 'Frete Médio'])

print("\nFrete Médio por Loja:")
print(resultado.to_string(index=False, float_format="%.2f"))  # Exibe com 2 casas decimais
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Carregar os dados
urls = {
    "Loja 1": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv",
    "Loja 2": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv",
    "Loja 3": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv",
    "Loja 4": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"
}

# Dicionário para armazenar os fretes médios
frete_medio = {}

# Função para detectar coluna de frete
def encontrar_coluna_frete(df):
    possiveis_nomes = ['Frete']
    for nome in possiveis_nomes:
        if nome in df.columns:
            return nome
    return None

# Calcular frete médio por loja
for nome_loja, url in urls.items():
    df = pd.read_csv(url)
    coluna = encontrar_coluna_frete(df)

    if coluna:
        # Converter para numérico se necessário (caso tenha símbolos como R$)
        df[coluna] = pd.to_numeric(df[coluna].astype(str).str.replace('[^\d.]', '', regex=True), errors='coerce')
        frete_medio[nome_loja] = df[coluna].mean()

# Criar DataFrame com os resultados
df_frete = pd.DataFrame(list(frete_medio.items()), columns=['Loja', 'Frete Médio'])

# Configurar o gráfico de pizza
plt.figure(figsize=(10, 8))
colors = ['#66b3ff','#99ff99','#ffcc99','#ff9999']
explode = (0.05, 0.05, 0.05, 0.05)  # Separar as fatias

# Plotar o pie chart
patches, texts, autotexts = plt.pie(
    df_frete['Frete Médio'],
    labels=df_frete['Loja'],
    autopct='%1.1f%%',
    startangle=90,
    colors=colors,
    explode=explode,
    shadow=True,
    textprops={'fontsize': 12}
)

# Melhorar a legenda
plt.legend(
    patches,
    [f"{loja} - R${frete:.2f}" for loja, frete in zip(df_frete['Loja'], df_frete['Frete Médio'])],
    title="Frete Médio",
    loc="center left",
    bbox_to_anchor=(1, 0, 0.5, 1)
)

# Ajustar título
plt.title('Distribuição do Frete Médio por Loja', fontsize=16, pad=20)

# Equal aspect ratio para que o pie seja desenhado como um círculo
plt.axis('equal')

plt.tight_layout()
plt.show()
```

#Atividade Extra: Análise de Desempenho Geográfico.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Carregar os dados
urls = {
    "Loja 1": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv",
    "Loja 2": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv",
    "Loja 3": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv",
    "Loja 4": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"
}

# Dados de faturamento
faturamento = {
    "Loja 1": 1534509.12,
    "Loja 2": 1488459.06,
    "Loja 3": 1464025.03,
    "Loja 4": 1384497.58
}

# Carregar e combinar os dados
dfs = []
for nome, url in urls.items():
    df = pd.read_csv(url)
    df['Loja'] = nome
    df['Faturamento_Loja'] = faturamento[nome]
    dfs.append(df)

dados_completos = pd.concat(dfs)
```

```python
plt.figure(figsize=(12, 8))
sns.scatterplot(data=dados_completos, x='lon', y='lat', hue='Loja',
                palette='viridis', alpha=0.6, size='Faturamento_Loja', sizes=(20, 200))
plt.title('Distribuição Geográfica das Vendas por Loja', fontsize=16)
plt.xlabel('Longitude', fontsize=12)
plt.ylabel('Latitude', fontsize=12)
plt.legend(title='Loja', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
